```python
import re
import nltk
import pandas as pd

from collections import Counter
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag
from nltk.corpus import wordnet

#ni bagian download nya
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')
nltk.download('punkt_tab')
```

```python
# ini corpus nya

corpus = """The new P is “Purple Cow.
When my family and I were driving through France a few
years ago, we were enchanted by the hundreds of storybook
cows grazing on picturesque pastures right next to the
highway. For dozens of kilometers, we all gazed out the
window, marveling about how beautiful everything was.
Then, within twenty minutes, we started ignoring the
cows. The new cows were just like the old cows, and what
once was amazing was now common. Worse than common.
It was boring.
Cows, after you’ve seen them for a while, are boring.
They may be perfect cows, attractive cows, cows with great
personalities, cows lit by beautiful light, but they’re still
boring.
A Purple Cow, though. Now that would be interesting.
(For a while.)
The essence of the Purple Cow is that it must be
remarkable."""

print(corpus)
```

```python
#1. Jumlah karakter corpus awal;

jumlah_karakter = len(corpus)
print(jumlah_karakter)
```

```python
# 2. Jumlah kata/token sebelum preprocessing;
nltk.download('punkt')
nltk.download('punkt_tab')
token = word_tokenize(corpus)
jumlah_token = len(token)
print(f"jumlah token awal: {jumlah_token}")
print(token[:20])
```

```python
# 3. Jumlah token unik;
token_unik_awal = set(token)
print(f"Jumlah token unik awal: {len(token_unik_awal)}")
print(list(token_unik_awal)[:20])
```

```python
#4. Jumlah token setelah stopword removal;

stop_words = set(stopwords.words('english'))

print("jumlah stopword bahasa inggris :" , len(stop_words))

print("\nContoh Stopword:")

print(list(stop_words)[:30])
```

```python
stop_words = set(stopwords.words('english'))

token = word_tokenize(corpus)

tokens_tanpa_stopword = [word for word in token if word.lower() not in stop_words]

print(f"jumlah token sebelum stopword removal:  {len(token)}")
print(f"jumlah token setelah stopword removal: {len(tokens_tanpa_stopword)}\n")

print("hasilan stopword removal: ")
print(tokens_tanpa_stopword[:20])

frekuensi_kata = Counter(tokens_tanpa_stopword)
top_10 = frekuensi_kata.most_common(10)

print("\n10 kata yang paling muncul: \n")

for kata, frekuensi_kata_val in top_10:
    print(f"{kata:<15}: {frekuensi_kata_val}")
```

```python
## 10 kata yang paling muncul

frekuensi_kata = Counter(tokens_tanpa_stopword)
top_10=frekuensi_kata.most_common(10)
print("10 kata yang paling muncul: \n")

for kata, frekuensi_kata_val in top_10:
    print(f"{kata:<15}: {frekuensi_kata_val}")
```

```python
# 6. Hasil stemming;

stemmer = PorterStemmer()
tokens_stemmed = [stemmer.stem(word) for word in tokens_tanpa_stopword]

print("Hasil stemming: ")
print(tokens_stemmed[:20])
```

```python
# 7. Hasil lemmatization;

lemmatizer = WordNetLemmatizer()
tokens_lemmatized = [lemmatizer.lemmatize(word) for word in tokens_tanpa_stopword]

print("lemmatization nya : ")
print(tokens_lemmatized[:20])
```

```python
#8. Perbedaan jumlah vocabulary sebelum dan sesudah preprocessing;
print("8. Perbedaan jumlah vocabulary sebelum dan sesudah preprocessing;")

jumlah_token_unik_awal = len(set(token))
jumlah_token_unik_setelah_stopword = len(set(tokens_tanpa_stopword))
jumlah_token_unik_setelah_stemming = len(set(tokens_stemmed))
jumlah_token_unik_setelah_lemmatization = len(set(tokens_lemmatized))

print(f"\nJumlah vocabulary awal: {jumlah_token_unik_awal}")
print(f"Jumlah vocabulary setelah stopword removal: {jumlah_token_unik_setelah_stopword}")
print(f"Jumlah vocabulary setelah stemming: {jumlah_token_unik_setelah_stemming}")
print(f"Jumlah vocabulary setelah lemmatization: {jumlah_token_unik_setelah_lemmatization}")
```

```python
# ### 9. Apakah preprocessing menghilangkan informasi penting?

# Preprocessing teks seperti *stopword removal*, *stemming*, dan *lemmatization* memang dapat menghilangkan beberapa 'informasi' dari data asli. Namun, 'informasi penting' ini sangat bergantung pada tujuan analisis Anda dan konteks data
# *   **Stopword Removal:** Menghilangkan kata-kata umum (seperti 'the', 'is', 'a') yang seringkali tidak membawa makna substantif dalam analisis teks. Ini biasanya mengurangi 'noise' dan fokus pada kata kunci yang lebih relevan. Jika nuansa bahasa atau keberadaan kata-kata umum itu sendiri penting (misalnya untuk analisis gaya bahasa), maka penghilangan ini bisa dianggap menghilangkan informasi.

# *   **Stemming & Lemmatization:** Mengubah kata menjadi bentuk dasar atau akar kata (misalnya, 'running', 'ran', 'runs' menjadi 'run'). Ini mereduksi jumlah variasi kata, membantu dalam mengelompokkan kata-kata yang memiliki makna serupa, dan mengurangi kompleksitas data. Namun, proses ini bisa menghilangkan detail morfologi kata yang mungkin penting untuk tugas-tasks tertentu, seperti analisis tata bahasa atau konteks yang sangat spesifik. Misalnya, 'universities' dan 'university' akan menjadi 'univers' setelah stemming, yang menghilangkan perbedaan jumlah.

# Secara umum, tujuan dari preprocessing adalah **mengurangi dimensi data dan fokus pada inti makna** untuk meningkatkan kinerja model atau analisis. Dalam banyak kasus, informasi yang dihilangkan selama preprocessing dianggap tidak relevan atau 'noise' untuk tugas yang sedang dikerjakan. Namun, jika detail linguistik atau keberadaan kata-kata umum sangat krusial, maka preprocessing harus dilakukan dengan hati-hati atau disesuaikan.
```

```python
#di import semuanya dulu set up nya
import re
import math
import numpy as np
import pandas as pd
from collections import Counter
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print ("semua libray oke selesai diintals")
```

```python
#
doc1 ="Why do birds fly in formation? Because the birds that follow the leader have an easier flight. The leader breaks the wind resistance, and the following birds can fly far more efficiently."
doc2 ="The problem with people who would avoid a remarkable career is that they never end up as the leader. They decide to work for a big company, intentionally functioning as an anonymous drone, staying way back to avoid risk and criticism."

print ("DOKUMEN 1 nya : (Halaman 65)")
print (doc1)
print ("DOKUMEN 2 nya : (Halaman 65)")
print (doc2)
```

```python
from sklearn import preprocessing
def preprocess(text):
  text = text.lower()
  text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
  tokens = text.split()
  return tokens

tokens1 = preprocess(doc1)
tokens2 = preprocess(doc2)

print("Token Dokumen 1:")
print(tokens1)
print("jumlah token dokumen 1: ")
print(len(tokens1))
print("Token Dokumen 2:")
print(tokens2)
print("jumlah token DOkumen 2: ")
print(len(tokens2))
```

```python
#kata apa saja yang muncul di kedua dokumen?
kata_bersama = sorted(set(tokens1)&set(tokens2))
print("Kata yang sering muncul ", kata_bersama)

```

```python
vocab = sorted(set(tokens1 + tokens2))
print(f"Vocalbulary nya : ({len(vocab)} kata unik):")
print(vocab)
```

```python
def one_hot(word, vocab):
  vec = [0] * len(vocab)
  if word in vocab :
    vec[vocab.index(word)] = 1
  else:
      print(f"kata {word} tidak ada di vocab")
  return vec

oh_leader = one_hot("leader", vocab)
oh_birds = one_hot("birds" , vocab)

print("One-hot 'leader':", oh_leader)
print("Index kata 'leader' pada vocabulary;", vocab.index("leader"))
print()
print("One-hot 'bird' :", oh_birds)
print("Index kata 'bird' pada vocabulary ", vocab.index("birds"))
```

```python
#Bigram nya
def get_ngrams(tokens, n):
  return tuple(tokens[i:i+n] for i in range(len(tokens) - n + 1))

bigrams1 = get_ngrams(tokens1, 2)
print(f"\nBigrams Dokumen 1:")
print(bigrams1)
T, N = len(tokens1), 2
print(f"\nVerisikasi rumus: Jumlah N-Gram = {T}-{N}+1= {T-N+1}")
print("Jumlah bigram aktual " , len(bigrams1))

```

```python
from os.path import join
corpus = [" " .join(tokens1), " ". join(tokens2)]

CV= CountVectorizer()
bow_matrix = CV.fit_transform(corpus)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=CV.get_feature_names_out(),
    index=["DOK 1", "DOK 2"]
)
print("Matriks Bag of Words: (baris = dokumen, kolom = kata) : ")
print(bow_df)
```

```python
# kata dengan frekuensi tertinggi di masing masig dokumen
for doc in bow_df.index:
  top3 = bow_df.loc[doc].sort_values(ascending=False).head(3)
  print(f"\nTop 3 kata untuk dokumen {doc}:")
  print(top3)
  print()
```

```python
c1, c2 = Counter(tokens1), Counter(tokens2)
N_docs = 2

kata_kunci = ['birds', 'leader', 'formation', 'remakable', 'avoid', 'the']
hasil = []
for w in kata_kunci:
  df = (w in c1) + (w in c2)
  idf = math.log((N_docs + 1) / (df + 1)) + 1
  tf1 = c1[w] / len(tokens1)
  tf2 = c2[w] / len(tokens2)
  hasil.append([w, round(tf1,3), round(tf2,3), df,round(idf,3),round(tf1*idf,4), round(tf2*idf,4)])

manual_df = pd.DataFrame(hasil, columns=['kata', 'tf1', 'tf2', 'df', 'idf', 'tf-idf1', 'tf-idf2'])
print(manual_df)
```

```python
tfidf_vec = TfidfVectorizer()
tfidf_matrix = tfidf_vec.fit_transform(corpus)

tfidf_df = pd.DataFrame(
  np.round(tfidf_matrix.toarray(),3),
  columns=tfidf_vec.get_feature_names_out(),
  index=["DOK 1", "DOK 2"]
)
print("Matriks TF-IDF:")
print(tfidf_df)
```

```python
cos_bow = cosine_similarity(bow_matrix[0], bow_matrix[1])[0][0]
cos_tfidf = cosine_similarity(tfidf_matrix[0], tfidf_matrix[1])[0][0]

print("Cosine Similarity berbasis BoW : ",round(cos_bow, 3))
print("Cosine Similarity berbasis TF-IDF : ",round(cos_tfidf, 3))
```

```python
#di import semuanya dulu set up nya
import re
import math
import numpy as np
import pandas as pd
from collections import Counter
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print ("semua libray oke selesai diintals")
```

```python
#Dataset :
#DOKUMEN 1 · Halaman 17
#DOKUMEN 2 · Halaman 16
doc1 ="Something remarkable is worth talking about. Worth noticing. Exceptional. New. Interesting. It's a Purple Cow. Boring stuff is invisible. It's a brown cow."
doc2 ="A Purple Cow, though. Now that would be interesting. The essence of the Purple Cow is that it must be remarkable."

print ("DOKUMEN 1 nya : (Halaman 65)")
print (doc1)
print ("DOKUMEN 2 nya : (Halaman 65)")
print (doc2)

#Catatan: kedua dokumen sengaja berbagi kata kunci (“Purple”, “Cow”, “remarkable”, “interesting”) agar perhitungan BoW, TF-IDF, dan Cosine Similarity menghasilkan pola yang bermakna untuk dianalisis.
```

```python
#SOAL 1
#Lakukan preprocessing dasar terhadap Dokumen 1 dan Dokumen 2:
def preprocess(text):
  #a. Case folding — ubah seluruh teks menjadi huruf kecil.
  text = text.lower()
  #b. Hilangkan tanda baca (punctuation) menggunakan regex.
  text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
  #c. Tokenisasi — pecah setiap dokumen menjadi daftar token/kata.
  tokens = text.split()
  return tokens

tokens1 = preprocess(doc1)
tokens2 = preprocess(doc2)

#Berapa jumlah token pada masing-masing dokumen setelah preprocessing? Apakah ada kata yang muncul di kedua dokumen?
print("Token Dokumen 1:")
print(tokens1)
print("jumlah token dokumen 1: ")
print(len(tokens1))
print("\nToken Dokumen 2:")
print(tokens2)
print("jumlah token Dokumen 2: ")
print(len(tokens2))

# Pertanyaan analisis: Apakah ada kata yang muncul di kedua dokumen?
common_words = set(tokens1).intersection(set(tokens2))

print("\nKata-kata yang muncul di kedua dokumen:")
if common_words:
    print(list(common_words))
else:
    print("Tidak ada kata yang muncul di kedua dokumen.")
```

```python
#SOAL 2
#a. Bangun vocabulary (kosakata unik) dari gabungan token kedua dokumen hasil Soal 1.
vocab = sorted(set(tokens1 + tokens2))
print(f"Vocalbulary nya : ({len(vocab)} kata unik):")
print(vocab)

#b. Tentukan panjang vektor one-hot (= jumlah kata unik pada vocabulary).
def one_hot(word, vocab):
  vec = [0] * len(vocab)
  if word in vocab :
    vec[vocab.index(word)] = 1
  else:
      print(f"\nkata {word} tidak ada di vocab")
  return vec

#c. Buat representasi one-hot encoding untuk kata “purple” dan kata “cow”.
oh_purple = one_hot("purple", vocab)
oh_cow = one_hot("cow" , vocab)

print("\nOne-hot 'purple': ", oh_purple)
print("Index kata 'purple' pada vocabulary;", vocab.index("purple"))
print()
print("One-hot 'cow' :", oh_cow)
print("Index kata 'cow' pada vocabulary ", vocab.index("cow"))

#analisis
print("\n\n### Analisis:")
print("Mengapa ukuran vektor one-hot akan membengkak jika vocabulary semakin besar?")
print("Ukuran vektor one-hot akan membengkak seiring dengan bertambahnya ukuran vocabulary karena setiap kata unik dalam vocabulary membutuhkan satu dimensi tersendiri dalam vektor. Jadi, jika ada N kata unik dalam vocabulary, setiap vektor one-hot akan memiliki panjang N. Ketika vocabulary tumbuh, N meningkat secara proporsional, menyebabkan vektor menjadi sangat panjang.")

print("\nApa kelemahan utama pendekatan ini?")
print("Kelemahan utama pendekatan one-hot encoding adalah:")
print("1. **Sparsity (Kepadatan Rendah)**: Sebagian besar elemen dalam vektor one-hot adalah nol (hanya satu '1' dan sisanya '0'). Hal ini menyebabkan penggunaan memori dan komputasi yang tidak efisien, terutama untuk vocabulary yang besar.")
print("2. **Tidak Menangkap Hubungan Semantik**: One-hot encoding memperlakukan setiap kata sebagai entitas yang sepenuhnya independen. Tidak ada informasi yang terkandung dalam vektor yang menunjukkan kesamaan atau hubungan semantik antara kata-kata (misalnya, 'raja' dan 'ratu' dianggap sama jauhnya satu sama lain dengan 'raja' dan 'pisang'). Ini menghambat kemampuan model untuk belajar dari konteks atau kesamaan makna antar kata.")
print("3. **Dimensionalitas Tinggi (Curse of Dimensionality)**: Dengan vocabulary yang besar, vektor one-hot menjadi sangat tinggi dimensinya. Ini dapat menyebabkan 'curse of dimensionality', di mana model memerlukan lebih banyak data pelatihan dan sumber daya komputasi untuk dapat belajar secara efektif, dan dapat mempersulit generalisasi model.")
```

```python
#SOAL 3
#a. Gunakan token hasil preprocessing Dokumen 1 (Soal 1).
tokens1 = preprocess(doc1)
print("Token Dokumen 1: ")
print(tokens1)
print("Jumlah token dokumen 1: ")
print(len(tokens1))

#b. Bentuk seluruh bigram (N=2) dari token tersebut.
def get_ngrams(tokens, n):
  return tuple(tokens[i:i+n] for i in range(len(tokens) - n + 1))

bigrams1 = get_ngrams(tokens1, 2)
print(f"\nBigrams Dokumen 1:")
print(bigrams1)

#c. Hitung jumlah bigram menggunakan rumus Jumlah N-Gram = T − N + 1, lalu cocokkan dengan hasil pada poin (b).
T, N = len(tokens1), 2
print(f"\nVerifikasi rumus: Jumlah N-Gram = {T}-{N}+1= {T-N+1}")
print("Jumlah bigram aktual ", len(bigrams1))

print("\n\n### Analisis:")
print("Bigram mana yang menurutmu paling menangkap makna “Purple Cow” sebagai satu konsep? Jelaskan.")
print("Bigram yang paling menangkap makna \"Purple Cow\" sebagai satu konsep adalah `('purple', 'cow')`.")
print("Penjelasan: Bigram ini secara langsung menggabungkan dua kata tersebut. Dalam konteks dokumen, 'Purple Cow' digunakan sebagai metafora untuk sesuatu yang 'remarkable', 'exceptional', atau 'interesting'. Ketika kedua kata ini muncul bersamaan, mereka membentuk satu unit semantik yang spesifik, berbeda dengan makna jika kata 'purple' dan 'cow' muncul secara terpisah. Ini menunjukkan bahwa keberadaan kedua kata ini secara berurutan membentuk konsep kunci yang dimaksud dalam teks.")
```

```python
#SOAL 4
#a. Gunakan CountVectorizer dari scikit-learn pada Dokumen 1 dan Dokumen 2 (hasil preprocessing).
corpus = [" ".join(tokens1), " ".join(tokens2)]
CV= CountVectorizer()
bow_matrix = CV.fit_transform(corpus)

#b. Tampilkan matriks frekuensi kata (baris = dokumen, kolom = kata pada vocabulary).
bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=CV.get_feature_names_out(),
    index=["DOK 1", "DOK 2"]
)
print("Matriks Bag of Words: (baris = dokumen, kolom = kata) : ")
print(bow_df)

#c. Identifikasi 3 kata dengan frekuensi tertinggi pada masing-masing dokumen.
for doc in bow_df.index:
  top3 = bow_df.loc[doc].sort_values(ascending=False).head(3)
  print(f"\nTop 3 kata untuk dokumen {doc}:")
  print(top3)
  print()

print("\n\n### Analisis:")
print("Apakah BoW memperhatikan urutan kata? Berikan contoh dari corpus yang menunjukkan keterbatasan tersebut.")
print("Tidak, Bag of Words (BoW) tidak memperhatikan urutan kata. BoW memperlakukan setiap dokumen sebagai 'kantong' kata-kata, di mana hanya frekuensi kemunculan setiap kata yang dihitung, tanpa mempertimbangkan posisi relatif atau urutan kata-kata tersebut.")
print("Keterbatasan ini terlihat jelas dalam contoh berikut dari dokumen kita:")
print("Dokumen 1 mengandung frasa 'Purple Cow' dan 'brown cow'.")
print("Dokumen 2 mengandung frasa 'Purple Cow'.")
print("Dalam matriks BoW, kita melihat hitungan terpisah untuk kata 'purple' dan 'cow'. Misalnya, di DOK 1, 'purple' muncul 1 kali dan 'cow' muncul 2 kali. Di DOK 2, 'purple' muncul 2 kali dan 'cow' muncul 2 kali.")
print("Model BoW tidak dapat membedakan antara 'Purple Cow' sebagai sebuah konsep unik (yang merupakan inti dari dokumen ini) dengan kemunculan kata 'cow' yang diikuti oleh kata lain, seperti 'brown cow'. Bagi BoW, 'purple' adalah sebuah kata, dan 'cow' adalah sebuah kata, dan ia hanya menghitung berapa kali masing-masing kata muncul, tanpa memahami bahwa 'purple cow' memiliki makna semantik yang spesifik sebagai satu kesatuan. Ini adalah kelemahan utama BoW, karena informasi kontekstual dan semantik yang penting seringkali terkandung dalam urutan kata-kata.")
```

```python
#SOAL 5
#a. Hitung TF-IDF untuk Dokumen 1 dan Dokumen 2 menggunakan TfidfVectorizer.
# First, join tokens back into strings for TfidfVectorizer
corpus_tfidf = [" ".join(tokens1), " ".join(tokens2)]

Tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = Tfidf_vectorizer.fit_transform(corpus_tfidf)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=Tfidf_vectorizer.get_feature_names_out(),
    index=["DOK 1", "DOK 2"]
)
print("Matriks TF-IDF:")
print(tfidf_df)

#b. Tampilkan bobot TF-IDF tertinggi pada masing-masing dokumen.
print("\nKata dengan bobot TF-IDF tertinggi per dokumen:")
for doc_idx, doc_name in enumerate(tfidf_df.index):
    top_tfidf_words = tfidf_df.iloc[doc_idx].sort_values(ascending=False).head(3)
    print(f"Untuk {doc_name}:")
    print(top_tfidf_words)
    print()

#c. Bandingkan bobot kata “remarkable” dan “cow” dengan hasil BoW pada Soal 4.
print("\n### Perbandingan Bobot 'remarkable' dan 'cow' (BoW vs TF-IDF):")
words_to_compare = ['remarkable', 'cow']

for word in words_to_compare:
    print(f"\nKata: '{word}'")
    if word in bow_df.columns:
        print(f"  BoW Frekuensi Dok 1: {bow_df.loc['DOK 1', word]}")
        print(f"  BoW Frekuensi Dok 2: {bow_df.loc['DOK 2', word]}")
    else:
        print(f"  Kata '{word}' tidak ditemukan dalam vocabulary BoW.")

    if word in tfidf_df.columns:
        print(f"  TF-IDF Bobot Dok 1: {tfidf_df.loc['DOK 1', word]:.4f}")
        print(f"  TF-IDF Bobot Dok 2: {tfidf_df.loc['DOK 2', word]:.4f}")
    else:
        print(f"  Kata '{word}' tidak ditemukan dalam vocabulary TF-IDF.")

print("\n**Observasi:**")
print("  - Untuk 'remarkable': BoW menunjukkan frekuensi 1 di kedua dokumen. TF-IDF juga memberikan bobot yang relatif tinggi karena kata ini penting dalam kedua dokumen dan tidak terlalu sering muncul di corpus keseluruhan.")
print("  - Untuk 'cow': BoW menunjukkan frekuensi 2 di kedua dokumen. Namun, bobot TF-IDF untuk 'cow' cenderung lebih rendah dibandingkan 'remarkable', terutama di DOK 1. Ini karena 'cow' muncul di kedua dokumen, yang mengurangi nilai IDF-nya, menandakan bahwa 'cow' kurang unik atau diskriminatif dibandingkan 'remarkable' dalam konteks kedua dokumen ini.")


#Pertanyaan analisis: Mengapa kata yang muncul di kedua dokumen memiliki bobot TF-IDF rendah (mendekati 0)? Kaitkan dengan konsep IDF.
print("\n\n### Analisis: Mengapa kata yang muncul di kedua dokumen memiliki bobot TF-IDF rendah (mendekati 0)?")
print("Kata yang muncul di kedua dokumen cenderung memiliki bobot TF-IDF rendah (atau lebih rendah dibandingkan kata unik lainnya) karena konsep **IDF (Inverse Document Frequency)**.")
print("IDF mengukur seberapa penting sebuah kata dalam seluruh korpus. IDF dihitung dengan rumus log((Jumlah Total Dokumen + 1) / (Jumlah Dokumen yang Mengandung Kata + 1)) + 1. Semakin banyak dokumen yang mengandung suatu kata, semakin rendah nilai IDF-nya (mendekati 1), dan sebaliknya.")
print("Jika sebuah kata muncul di **semua** dokumen dalam korpus (dalam kasus kita, jika sebuah kata muncul di kedua DOK 1 dan DOK 2), maka: Jumlah Dokumen yang Mengandung Kata = Jumlah Total Dokumen. Ini akan membuat rasio (Jumlah Total Dokumen + 1) / (Jumlah Dokumen yang Mengandung Kata + 1) mendekati 1. Logaritma dari 1 adalah 0. Sehingga, nilai IDF akan mendekati 1 (karena penambahan 1 di akhir rumus).")
print("Karena TF-IDF adalah hasil perkalian Term Frequency (TF) dan IDF, maka nilai IDF yang mendekati 1 akan menyebabkan bobot TF-IDF menjadi relatif rendah (terutama jika TF-nya juga tidak terlalu tinggi). Kata-kata semacam ini dianggap kurang informatif atau diskriminatif karena mereka muncul secara merata di seluruh korpus, sehingga tidak banyak membantu membedakan satu dokumen dari yang lain.")
print("Contoh kata seperti 'is' dan 'a' seringkali memiliki bobot TF-IDF yang sangat rendah karena mereka adalah kata umum (stopwords) yang muncul di hampir setiap dokumen, sehingga IDF mereka sangat rendah.")
```

```python
#SOAL 6

#a. Hitung Cosine Similarity antara vektor BoW Dokumen 1 dan Dokumen 2 (Soal 4).
cos_bow = cosine_similarity(bow_matrix[0].reshape(1, -1), bow_matrix[1].reshape(1, -1))[0][0]
print("Cosine Similarity berbasis BoW : ",round(cos_bow, 3))

#b. Hitung Cosine Similarity antara vektor TF-IDF Dokumen 1 dan Dokumen 2 (Soal 5).
cos_tfidf = cosine_similarity(tfidf_matrix[0].reshape(1, -1), tfidf_matrix[1].reshape(1, -1))[0][0]
print("Cosine Similarity berbasis TF-IDF : ",round(cos_tfidf, 3))

#c. Bandingkan kedua nilai tersebut.
print("\n### Perbandingan Cosine Similarity:")
print(f"Nilai Cosine Similarity BoW ({round(cos_bow, 3)}) lebih tinggi dibandingkan TF-IDF ({round(cos_tfidf, 3)}).")
print("Ini menunjukkan bahwa berdasarkan frekuensi kemunculan kata saja (BoW), dokumen-dokumen tersebut terlihat lebih mirip. Namun, ketika mempertimbangkan bobot kata (TF-IDF), kemiripan sedikit menurun karena kata-kata umum kurang diberi bobot.")

#Pertanyaan analisis: Berdasarkan nilai cosine similarity yang diperoleh, apakah Dokumen 1 dan Dokumen 2 dapat dikatakan “cukup mirip”? Kaitkan dengan isi kedua dokumen.
print("\n\n### Analisis: Apakah Dokumen 1 dan Dokumen 2 dapat dikatakan 'cukup mirip'?")
print("Berdasarkan nilai cosine similarity yang diperoleh:")
print(f"- **BoW: {round(cos_bow, 3)}**")
print(f"- **TF-IDF: {round(cos_tfidf, 3)}**")
print("\nNilai cosine similarity berkisar antara 0 (tidak mirip sama sekali) hingga 1 (sangat mirip/identik). Umumnya, nilai di atas 0.5 sering dianggap 'cukup mirip', meskipun ambang batas ini bisa bervariasi tergantung konteks.")
print("Dalam kasus ini, kedua nilai (sekitar 0.34 dan 0.21) relatif rendah, menunjukkan bahwa Dokumen 1 dan Dokumen 2 **tidak dapat dikatakan 'cukup mirip'** dalam arti memiliki kesamaan yang kuat atau substansial.")
print("\n**Kaitan dengan isi kedua dokumen:**")
print("Meskipun kedua dokumen membahas konsep 'Purple Cow' dan berbagi beberapa kata kunci ('purple', 'cow', 'remarkable', 'interesting'), Dokumen 1 secara umum lebih banyak memperkenalkan konsep 'Purple Cow' sebagai sesuatu yang 'remarkable', 'exceptional', dan membandingkannya dengan 'brown cow' yang 'boring'. Dokumen 2 lebih fokus pada esensi dari 'Purple Cow' itu sendiri.")
print("Perbedaan konteks dan penekanan ini, meskipun menggunakan beberapa kata kunci yang sama, menyebabkan vektor representasi mereka (baik BoW maupun TF-IDF) tidak terlalu sejajar. TF-IDF bahkan memberikan nilai similarity yang lebih rendah karena berhasil mengurangi bobot kata-kata yang umum di kedua dokumen, sehingga lebih menyoroti perbedaan unik antar dokumen.")
print("Dengan demikian, nilai cosine similarity yang rendah ini secara akurat mencerminkan bahwa meskipun ada benang merah topik, struktur kalimat, kedalaman pembahasan, dan frekuensi kata-kata non-kunci yang berbeda membuat kedua dokumen tidak memiliki kemiripan yang tinggi.")
```
